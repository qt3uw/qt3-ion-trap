```python
variac_volts = [2*n for n in range(13)]
variac_labels = list(map(str, variac_volts))
labels_v = [r'$\mathrm{{V}}_\mathrm{{est}}\pm \delta \mathrm{{V}}_\mathrm{{err}}$'.format(label = label) for label in variac_labels]
est_err = [[wid.FloatText(placeholder = r'Estimated', continuous_update = True), wid.FloatText(placeholder = r'Absolute Error', continuous_update = True)] for i in range(len(variac_labels))]
est_err_con = [wid.widget_bool.ToggleButton(placeholder = 'Confirm Masurements for' + variac_labels[i] + ' V', layout=Layout(width='90%')) for i in range(len(variac_labels))]
t_boxes_v = [VBox([HBox([wid.widget_string.Label(r'$\mathrm{{V}}_\mathrm{{est}}\pm \delta \mathrm{{V}}_\mathrm{{err}}:\ \ \ $   '), wid.widget_string.Label(r'$\Big($'), est_err[i][0], wid.widget_string.Label(r'$\pm \delta$'), est_err[i][1], wid.widget_string.Label(r'$\Big)\ \mathrm{V}$')]), HBox([wid.widget_bool.Valid(est_err_con[i].value), est_err_con[i]])]) for i in range(len(variac_labels))]
[wid.widget_link.jsdlink((est_err_con[i], 'value'), (est_err[j][i], 'disabled')) for i, j in zip(range(len(variac_labels)), range(2))]
tab = wid.Tab()
tab.children = t_boxes_v
for i in range(len(labels_v)):
    tab.set_title(i, variac_labels[i] + ' V')
display(tab)
display(wid.IntProgress(bar_color='blue'))

```

```python
copy = est_err
scope_volts = []
applied_voltages = []
for i in range(len(copy)):
    scope_volts.append(copy[i][0].value)
    applied_voltages.append(copy[i][0].value * 10**3)

print(scope_volts)
print(applied_voltages)

# for child in tab.children:
#     for chilld in child.children:
#         scope_volts.append(chilld)
# print(scope_volts)
```
